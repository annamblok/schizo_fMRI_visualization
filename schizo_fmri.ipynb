Step 1 - The code cell below locates the MRI data files automatically based on the file
extensions and file names. The current working directory contains the folder to the MRI data.

```python
# loading the packages
import os
import glob

# seeing if i am in the right working directory
print(os.getcwd())

# folder where the fmri files are stored
source = "./schizo_ass"

# file extension for the fmri data
ext = "*.nii.gz"

# putting the file path together
file_path = os.path.join(source, ext)
print(file_path)

# detecting all the files with the .nii.gz extension within my folder
fmri_files = glob.glob(file_path)
print(fmri_files)
```

Step 2 - The code cell below allows for visualization of the functional data on top of the anatomical scan using nilearn’s plot_stat_map function. 




```python
# import the right packages
from nilearn import plotting

# create a variable name that links to the right nifti images
stat_map_img = "./schizo_ass\\schizophrenia_uniformity-test_z_FDR_0.01.nii.gz"
bg_img = "./schizo_ass\\anatomical (1).nii.gz"

# use the plot_stat_map function to plot the figure
plotting.plot_stat_map(
    stat_map_img,           
    bg_img,                 
    cut_coords=(36, 24, -2), 
    title="Schizophrenia-related fMRI activation",
    threshold = 0.1,
    draw_cross = False,
    cmap = "plasma"    
)
```

Step 3 - The code cell below plots a histogram of the values in the functional data file. The functional data is loaded using nibabel, the positive values are selected, and then a histogram is plotted. 

```python
# import the libraries that are needed
import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt

# load the nifti file using nibabel
func_info = nib.load(stat_map_img)

# get the actual functional data points
func_data = func_info.get_fdata()

# filter the data points to keep positive values
pos_values = func_data > 0

# store these values 
pos_funcdata = func_data[pos_values]

# plot the histogram
plt.hist(pos_funcdata, bins = 40, color = "lightseagreen")
plt.xlabel("Voxel intensity")
plt.ylabel("Frequency")
plt.title("Schizophrenia-related fMRI voxel intensity distribution")

```
